## Energy Monitoring System

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
```

```python
names = ['timestamp', 'electricity_consumption', 'temperature', 'humidity', 'wind_speed', 'avg_past_consumption', 'class']
dataset = pd.read_csv('C:/Users/bisht/Downloads/smart_meter_data.csv', names = names, skiprows = 1)
```

```python
dataset.drop(columns = ['timestamp'], inplace = True)
```

```python
dataset.sample(5)
```

```python
dataset['class'].unique()
```

```python
arr = dataset.values
X = arr[:, 0:-1]
y = arr[:, -1]

# print(X[25])
# print(y[25])

X_train, X_val, y_train, y_val = train_test_split(X, y, test_size = 0.3, random_state = 1)
```

```python
model = SVC(gamma = 'auto')
model.fit(X_train, y_train)

result = model.score(X_val, y_val)
print(result)
```

```python
value = [[212.7315, 0.84753, 180.2963282, 32.25364, 4.364337]]
preds = model.predict(value)
print(preds[0])
```
